__7. Refering to the below model, describe underfitting in Linear Regression.__

````python
x_train = np.array([1, 1.5, 1.8, 2, 2.2])
y_train = np.array([[200], [250], [260], [280], [500]])
tmp_f_wb = compute_model_output(x_train, 200, 10,)
plt.plot(x_train, tmp_f_wb, c='b',label='Our Prediction')
plt.scatter(x_train, y_train, marker='x', c='r',label='Actual Values')
plt.show()
````


Under-fitting is where the line of best fit is too generalized and creates a lot of prediction error across a large amount of the dataset. This tends to be an issue with the model's hypothesis being too simplistic and which may miss the overall trend within the dataset's complexity.
