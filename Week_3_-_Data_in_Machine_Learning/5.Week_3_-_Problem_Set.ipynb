__2. In a markdown box, explain 'underfitting' using the below graph as a reference.__

```python
fig, ax = plt.subplots()
x = np.linspace(10, -10, 100)
y = -x**2 + 4*x + 2  
x = x + np.random.uniform(-3, 3, size=x.shape)
y = y + np.random.uniform(-5, 5, size=y.shape)
ax.scatter(x, y, c='b', label='Training & Testing data')
x_lin = np.array([-10,10])
tmp_f_mb = compute_model_output(x_lin, 0, 0,)
ax.plot(x_lin, tmp_f_mb, c='r', label='ML Model')
ax.set_xticks([])
ax.set_xticklabels([])
ax.set_yticks([])
ax.set_yticklabels([])
plt.legend()
plt.show()
```

Underfitting is when a model over generalises data, meaning that the fit doesn't catch the trend leading to high bias and loss of each individual point. If there is high loss for the training set then there will be high loss for predicted data. Underfitting can be due to the model being unsuitable for the target variable or is too simplistic and requires a higher index in order to create the appropriate fit.
